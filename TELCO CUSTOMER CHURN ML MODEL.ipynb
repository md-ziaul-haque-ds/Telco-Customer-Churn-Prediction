# 1. Load the CSV

In our Jupyter Notebook, load the Telco CSV into a DataFrame.

```python
import pandas as pd

df = pd.read_csv("telco_customer_churn.csv")
df
```

```python
# First 5 rows

df.head()
```

```python
# Shape

df.shape
```

```python
# Column names
df.columns
```

```python
# Data types

df.dtypes
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
# Data types
df.dtypes
```

```python
df.isnull().sum()
```

```python
df[df["TotalCharges"].isna()][["tenure", "MonthlyCharges", "TotalCharges", "Churn"]]
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df["TotalCharges"].isna().sum()
```

```python
# Train/Test Split
```

```python
X = df.drop(columns=["customerID", "Churn"])
y = df["Churn"]
```

```python
# Quickly verify:
print(X.shape)
print(y.shape)
```

```python
print(X.columns)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# And check the churn proportions:
```

```python
print("Train:")
print(y_train.value_counts(normalize=True))

print("\nTest:")
print(y_test.value_counts(normalize=True))
```

```python
X_train.dtypes
```

```python
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features = X_train.select_dtypes(
    include=["object", "str"]
).columns

print("Numeric:", numeric_features.tolist())
print("Categorical:", categorical_features.tolist())
```

```python
# Create preprocessing pipeline
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
# Build the ML Pipeline
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])
```

```python
# Now we train it 
```

```python
logistic_model.fit(X_train, y_train)
```

```python
# Make predictions
```

```python
y_pred = logistic_model.predict(X_test)
```

```python
y_pred[:20]
```

```python
# Also get churn probabilities
```

```python
y_proba = logistic_model.predict_proba(X_test)[:, 1]
```

```python
logistic_model.classes_
```

```python
# Classification Report
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Confusion Matrix
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=logistic_model.classes_
).plot()
```

```python
# metric: ROC-AUC.
```

```python
from sklearn.metrics import roc_auc_score

auc = roc_auc_score(y_test, y_proba)

print("ROC-AUC:", auc)
```

```python
from sklearn.metrics import RocCurveDisplay

RocCurveDisplay.from_predictions(
    y_test,
    y_proba,
    pos_label="Yes"
)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
import numpy as np
for threshold in [0.5, 0.4, 0.3, 0.2]:
    y_pred_threshold = np.where(
        y_proba >= threshold,
        "Yes",
        "No"
    )

    print(f"Threshold: {threshold}")
    print("Precision:", precision_score(y_test, y_pred_threshold, pos_label="Yes"))
    print("Recall:", recall_score(y_test, y_pred_threshold, pos_label="Yes"))
    print("F1:", f1_score(y_test, y_pred_threshold, pos_label="Yes"))
    print()
```

# Rndom Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)
y_proba_rf = rf_model.predict_proba(X_test)[:, 1]
```

```python
# Evaluate
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))

print("ROC-AUC:", roc_auc_score(y_test, y_proba_rf))
```

```python
# improve Logistic Regression properly
# Start by creating a balanced Logistic Regression pipeline:
```

```python
balanced_logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])
```

```python
balanced_logistic_model.fit(X_train, y_train)

y_pred_balanced = balanced_logistic_model.predict(X_test)
y_proba_balanced = balanced_logistic_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_balanced))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_balanced))

print("ROC-AUC:", roc_auc_score(y_test, y_proba_balanced))
```

```python
for threshold in [0.5, 0.4, 0.3, 0.2]:
    y_pred_threshold = np.where(
        y_proba_balanced >= threshold,
        "Yes",
        "No"
    )

    print(f"Threshold: {threshold}")
    print("Precision:", precision_score(
        y_test, y_pred_threshold, pos_label="Yes"
    ))
    print("Recall:", recall_score(
        y_test, y_pred_threshold, pos_label="Yes"
    ))
    print("F1:", f1_score(
        y_test, y_pred_threshold, pos_label="Yes"
    ))
    print()
```

```python
# Hyperparameter tuning
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "classifier__C": [0.01, 0.1, 1, 10, 100],
    "classifier__class_weight": [None, "balanced"],
    "classifier__solver": ["liblinear", "lbfgs"]
}
```

```python
grid_search = GridSearchCV(
    estimator=logistic_model,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=5,
    n_jobs=-1,
    verbose=1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best parameters:")
print(grid_search.best_params_)

print("\nBest CV score:")
print(grid_search.best_score_)
```

```python
# Evaluate the tuned model
```

```python
best_model = grid_search.best_estimator_

y_pred_tuned = best_model.predict(X_test)
y_proba_tuned = best_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tuned))

print("ROC-AUC:", roc_auc_score(y_test, y_proba_tuned))
```
